## PPO hoạt động như thế nào
1. Rollout: mô hình ngôn ngữ sinh phản hồi dựa trên query.
2. Evaluation: query và phản hồi được đánh giá bằng hàm, mô hình, phản hồi của con người hoặc một tổ hợp nào đó của chúng. Quá trình này phải sinh ra một **giá trị vô hướng** (scalar value) cho mỗi cặp query/phản hồi.
3. Optimization: ở bước tối ưu, cặp query/phản hồi được dùng để tính log-probability của các token trong chuỗi. Việc này được thực hiện bằng mô hình đang huấn luyện và mô hình tham chiếu. KL divergence giữa hai đầu ra được dùng làm tín hiệu thưởng bổ sung, nhằm bảo đảm phản hồi sinh ra không lệch quá xa mô hình ngôn ngữ tham chiếu. Sau đó dùng PPO để huấn luyện mô hình ngôn ngữ chủ động.

````python
config = PPOConfig(
    model_name="model/gpt2-imdb",
    learning_rate=1.41e-5,
    log_with="wandb",
)

sent_kwargs = {"top_k": None, "function_to_apply": "none", "batch_size": 16}
````

## Tải dữ liệu và mô hình

### Tải bộ dữ liệu IMDB  
Bộ dữ liệu IMDB gồm 50.000 bình luận phim, được gán nhãn cảm xúc "tích cực"/"tiêu cực". Chúng ta nạp bộ dữ liệu IMDB vào một DataFrame và lọc ra những bình luận dài ít nhất 200 ký tự. Sau đó, ta tiến hành tách từ (tokenize) từng đoạn văn bản và dùng `LengthSampler` để cắt ngẫu nhiên về độ dài chỉ định.

````python
def build_dataset(
    config,
    dataset_name="dataset/imdb",
    input_min_text_length=2,
    input_max_text_length=8,
):
    """
    Xây dựng bộ dữ liệu để huấn luyện. Hàm này tạo bộ dữ liệu từ `load_dataset`; người dùng nên
    tùy biến hàm này để huấn luyện mô hình trên bộ dữ liệu của riêng mình.

    Args:
        dataset_name (`str`):
            Tên của bộ dữ liệu cần nạp.

    Returns:
        dataloader (`torch.utils.data.DataLoader`):
            DataLoader cho bộ dữ liệu.
    """
    tokenizer = AutoTokenizer.from_pretrained(config.model_name)
    tokenizer.pad_token = tokenizer.eos_token
    # nạp imdb bằng datasets
    ds = load_dataset(dataset_name, split="train")
    ds = ds.rename_columns({"text": "review"})
    ds = ds.filter(lambda x: len(x["review"]) > 200, batched=False)

    input_size = LengthSampler(input_min_text_length, input_max_text_length)

    def tokenize(sample):
        sample["input_ids"] = tokenizer.encode(sample["review"])[: input_size()]
        sample["query"] = tokenizer.decode(sample["input_ids"])
        return sample

    ds = ds.map(tokenize, batched=False)
    ds.set_format(type="torch")
    return ds
````

````python
dataset = build_dataset(config)


def collator(data):
    return dict((key, [d[key] for d in data]) for key in data[0])
````

### Tải mô hình ngôn ngữ GPT2 đã huấn luyện trước
Chúng ta nạp mô hình GPT2 có value head và tokenizer. Mô hình được nạp hai lần; mô hình thứ nhất dùng cho việc tối ưu, mô hình thứ hai làm tham chiếu để tính KL divergence so với điểm khởi đầu. Đây là tín hiệu thưởng bổ sung trong huấn luyện PPO, nhằm bảo đảm mô hình sau tối ưu không lệch quá xa mô hình ngôn ngữ gốc.

````python
model = AutoModelForCausalLMWithValueHead.from_pretrained(config.model_name)
ref_model = AutoModelForCausalLMWithValueHead.from_pretrained(config.model_name)
tokenizer = AutoTokenizer.from_pretrained(config.model_name)

tokenizer.pad_token = tokenizer.eos_token
````

### Khởi tạo PPOTrainer  
`PPOTrainer` chịu trách nhiệm phân bổ thiết bị và thực hiện tối ưu về sau:

````python
ppo_trainer = PPOTrainer(
    config, model, ref_model, tokenizer, dataset=dataset, data_collator=collator
)
````

### Tải bộ phân loại BERT  
Chúng ta nạp một bộ phân loại BERT đã được tinh chỉnh trên bộ dữ liệu IMDB.

````python
device = ppo_trainer.accelerator.device
if ppo_trainer.accelerator.num_processes == 1:
    device = 0 if torch.cuda.is_available() else "cpu"  # để tránh lỗi của `pipeline`
sentiment_pipe = pipeline(
    "sentiment-analysis", model="model/distilbert-imdb", device=device
)
````

Mô hình xuất ra logits của lớp tiêu cực và lớp tích cực. Chúng ta sẽ dùng logits của lớp tích cực làm tín hiệu thưởng cho mô hình ngôn ngữ.

````python
text = "this movie was really bad!!"
sentiment_pipe(text, **sent_kwargs)
````

````python
text = "this movie was really good!!"
sentiment_pipe(text, **sent_kwargs)
````

### Thiết lập sinh  
Với việc sinh phản hồi, chúng ta chỉ dùng phương pháp lấy mẫu, đồng thời tắt top-k và nucleus sampling, và đặt một độ dài tối thiểu.

````python
gen_kwargs = {
    "min_length": -1,
    "top_k": 0.0,
    "top_p": 1.0,
    "do_sample": True,
    "pad_token_id": tokenizer.eos_token_id,
}
````

## Tối ưu mô hình

### Vòng lặp huấn luyện

Vòng lặp huấn luyện gồm các bước chính sau:
1. Lấy phản hồi cho query từ mạng chính sách (GPT-2)  
2. Lấy cảm xúc của cặp query/phản hồi từ BERT  
3. Tối ưu chính sách bằng PPO, dựa trên bộ ba (query, phản hồi, thưởng)  
<!-- **Thời gian huấn luyện**  
Với thiết lập nêu trên, bước này cần **khoảng 2 giờ** trên GPU V100. -->

````python
output_min_length = 4
output_max_length = 16
output_length_sampler = LengthSampler(output_min_length, output_max_length)


generation_kwargs = {
    "min_length": -1,
    "top_k": 0.0,
    "top_p": 1.0,
    "do_sample": True,
    "pad_token_id": tokenizer.eos_token_id,
}


for epoch, batch in enumerate(tqdm(ppo_trainer.dataloader)):
    query_tensors = batch["input_ids"]

    #### Lấy phản hồi từ gpt2
    response_tensors = []
    for query in query_tensors:
        gen_len = output_length_sampler()
        generation_kwargs["max_new_tokens"] = gen_len
        query_response = ppo_trainer.generate(query, **generation_kwargs).squeeze()
        response_len = len(query_response) - len(query)
        response_tensors.append(query_response[-response_len:])
    batch["response"] = [tokenizer.decode(r.squeeze()) for r in response_tensors]

    #### Tính điểm cảm xúc
    texts = [q + r for q, r in zip(batch["query"], batch["response"])]
    pipe_outputs = sentiment_pipe(texts, **sent_kwargs)
    positive_scores = [
        item["score"]
        for output in pipe_outputs
        for item in output
        if item["label"] == "POSITIVE"
    ]
    rewards = [torch.tensor(score) for score in positive_scores]

    #### Chạy một bước PPO
    stats = ppo_trainer.step(query_tensors, response_tensors, rewards)
    ppo_trainer.log_stats(stats, batch, rewards)
````

## Kiểm tra mô hình  
Hãy xem xét một vài ví dụ trong bộ dữ liệu IMDB. Chúng ta có thể dùng `ref_model` để so sánh mô hình sau tối ưu `model` với mô hình trước khi tối ưu.

````python
#### lấy một batch từ bộ dữ liệu
bs = 16
game_data = dict()
dataset.set_format("pandas")
df_batch = dataset[:].sample(bs)
game_data["query"] = df_batch["query"].tolist()
query_tensors = df_batch["input_ids"].tolist()

response_tensors_ref, response_tensors = [], []

#### lấy phản hồi từ gpt2 và gpt2_ref
for i in range(bs):
    query = torch.tensor(query_tensors[i]).to(device)

    gen_len = output_length_sampler()
    query_response = ref_model.generate(
        query.unsqueeze(0), max_new_tokens=gen_len, **gen_kwargs
    ).squeeze()
    response_len = len(query_response) - len(query)
    response_tensors_ref.append(query_response[-response_len:])

    query_response = model.generate(
        query.unsqueeze(0), max_new_tokens=gen_len, **gen_kwargs
    ).squeeze()
    response_len = len(query_response) - len(query)
    response_tensors.append(query_response[-response_len:])

#### giải mã các phản hồi
game_data["response (before)"] = [
    tokenizer.decode(response_tensors_ref[i]) for i in range(bs)
]
game_data["response (after)"] = [
    tokenizer.decode(response_tensors[i]) for i in range(bs)
]

#### phân tích cảm xúc của cặp query/phản hồi trước/sau
texts = [q + r for q, r in zip(game_data["query"], game_data["response (before)"])]
pipe_outputs = sentiment_pipe(texts, **sent_kwargs)
positive_scores = [
    item["score"]
    for output in pipe_outputs
    for item in output
    if item["label"] == "POSITIVE"
]
game_data["rewards (before)"] = positive_scores

texts = [q + r for q, r in zip(game_data["query"], game_data["response (after)"])]
pipe_outputs = sentiment_pipe(texts, **sent_kwargs)
positive_scores = [
    item["score"]
    for output in pipe_outputs
    for item in output
    if item["label"] == "POSITIVE"
]
game_data["rewards (after)"] = positive_scores

# lưu kết quả vào một dataframe
df_results = pd.DataFrame(game_data)
df_results
````

Quan sát giá trị trung bình/trung vị của thưởng trên các chuỗi được sinh ra, chúng ta thấy sự khác biệt rõ rệt.

````python
print("mean:")
display(df_results[["rewards (before)", "rewards (after)"]].mean())
print()
print("median:")
display(df_results[["rewards (before)", "rewards (after)"]].median())
````
